```python
df = pd.read_csv("CodeAlpha_WebScraping_Cleaned.csv")

print("Dataset loaded successfully!")

df.head()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='Rating')

plt.title('Distribution of Book Ratings', fontsize=16)
plt.xlabel('Book Rating (Stars)', fontsize=12)
plt.ylabel('Number of Books', fontsize=12)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

ax = sns.countplot(data=df, x='Rating')

for container in ax.containers:
    ax.bar_label(container)

plt.title('Book Ratings with Count', fontsize=16)
plt.xlabel('Book Rating (Stars)')
plt.ylabel('Number of Books')

plt.show()
```

```python
plt.figure(figsize=(10, 5))

sns.histplot(data=df, x='Price', bins=15, kde=True)

plt.title('Distribution of Book Prices', fontsize=16)
plt.xlabel('Book Price', fontsize=12)
plt.ylabel('Number of Books', fontsize=12)

plt.show()
```

```python
average_price = df['Price'].mean()

plt.figure(figsize=(7, 4))

sns.barplot(x=['Average Book Price'], y=[average_price])

plt.title('Average Price of Books', fontsize=16)
plt.ylabel('Price')

plt.show()

print("Average Book Price:", round(average_price, 2))
```

```python
plt.figure(figsize=(9, 5))

sns.boxplot(data=df, x='Rating', y='Price')

plt.title('Book Price Comparison by Rating', fontsize=16)
plt.xlabel('Book Rating (Stars)')
plt.ylabel('Book Price')

plt.show()
```

```python
rating_price = df.groupby('Rating')['Price'].mean().reset_index()

plt.figure(figsize=(9, 5))

sns.barplot(data=rating_price, x='Rating', y='Price')

plt.title('Average Book Price by Rating', fontsize=16)
plt.xlabel('Book Rating (Stars)')
plt.ylabel('Average Price')

plt.show()

print(rating_price)
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='Availability')

plt.title('Book Availability Analysis', fontsize=16)
plt.xlabel('Availability Status')
plt.ylabel('Number of Books')

plt.show()

print(df['Availability'].value_counts())
```

```python
availability_counts = df['Availability'].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    availability_counts,
    labels=availability_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Percentage of Book Availability')

plt.show()
```

```python
plt.figure(figsize=(7, 5))

correlation = df[['Price', 'Rating']].corr()

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Book Data')

plt.show()
```
